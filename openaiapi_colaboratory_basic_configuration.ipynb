# Basic prompt function

The **`get_completion`** function takes a **`prompt`** as input and generates a completion using the OpenAI library. It uses the provided prompt, model, and temperature to create a chat-based completion. The function returns the content of the completion generated.

**Parameters:**

| Parameter | Description |
| --- | --- |
| prompt | The input prompt for which the completion is generated. |
| model | The OpenAI model to use for generating the completion. Defaults to "gpt-3.5-turbo-16k-0613". |
| temperature | The temperature parameter controls the randomness of the completion. A higher value (e.g., 0.8) makes the output more random, while a lower value (e.g., 0.2) makes it more focused and deterministic. Defaults to 0. |



> 💡 By specifying the prompt, model, and temperature, you can fine-tune the behavior of the completion generation according to your requirements.

````python
# new
from openai import AsyncOpenAI

client = AsyncOpenAI()
completion = await client.chat.completions.create(model="gpt-3.5-turbo", messages=[{"role": "user", "content": "Hello world"}])
print (completion.choices[0].message.content)
async def get_completion(prompt, model="gpt-3.5-turbo", **kwargs):
    response = await client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        **kwargs
    )
    return response.choices[0].message.content
````

# Prompt Parameters Explained

## **Tokens**

Models read input as tokens (subwords/words)

Example: *"Hello world"* → ["Hello", " world"] = 2 tokens

Token count affects cost and length limits

````python
from transformers import GPT2Tokenizer
text = "Hello world!"
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
print(tokenizer.tokenize(text))
````

## **Temperature**

Controls randomness/creativity

* Low = safe/predictable
* High = creative

````python
prompt = "Write a poem about the moon."

result0 = await get_completion(prompt, temperature=0.2)
print(result0)
````

## **max_tokens**

Limits output length

````python
result = await get_completion("Describe the sun.", max_tokens=10)
print(result)
````

## **Top-p (nucleus sampling)**

Chooses from top tokens that sum to probability p

````python
result = await get_completion("Suggest a recipe with eggs.", top_p=0.5)
print(result)
````

## **Frequency_penalty & Presence_penalty**

Help reduce repetition in the output

**Frequency:** penalizes tokens that appear often

**Presence:** penalizes if a token already appeared at all

````python
result = await get_completion("Repeat the word 'innovation' in different sentences.",
              frequency_penalty=1.0,
              presence_penalty=0.5)
print(result)
````

# **Prompting**

## **Zero-shot Prompting**

**Definition:** No examples, only instructions

**Strengths:** Fast, simple

**Limitations:** Can be inaccurate for complex tasks

**Example Use Case:** Text classification

````python
prompt = f"""Classify the sentiment: 'I love this phone.'
        """
response = await get_completion(prompt)
print(response)
````

## **One-shot Prompting**

**Definition:** One example included

**Strengths:** Provides model with response format

**Limitations:** May overfit to example

**Example Use Case:** Text classification with template

````python
prompt = f""" Text: 'This is terrible.' Sentiment: Negative
              Text: 'Great service!' Sentiment:
        """
response = await get_completion(prompt)
print(response)
````

**Few-shot Prompting**

**Definition:** Multiple examples

**Strengths:** Improves task understanding (in-context learning)

**Best For:** Creative tasks, novel words, ambiguous tasks

**Example:** Inserting invented word into a sentence

````python
prompt = f""" A 'rose' is a flower. Example: 'The rose bloomed'
            Use 'rose' in a sentence:
        """
response = await get_completion(prompt)
print(response)
````

## **Chain-of-Thought (CoT)**

**Definition:** Prompting model to reason step-by-step

**Improves:** Math, logic, reasoning tasks

````python
prompt = f""" How many legs do 5 spiders have? Explain step by step.
        """
response = await get_completion(prompt)
print(response)
````
